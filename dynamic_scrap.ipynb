```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By

import pandas as pd
import time
```

```python
service = Service("C:/Users/asus/chromedriver-win64/chromedriver.exe")
options = Options()

options.add_argument("--start-maximized")
# options.add_argument("--headless-new") # open browser in background

driver = webdriver.Chrome(service=service, options=options)
time.sleep(4)

keywords = ['watch', 'laptop', 'shoes']
all_data = []

for i in keywords:

    url = f"https://www.amazon.in/s?k=watch"
    driver.get(url)

    products = driver.find_elements(By.CSS_SELECTOR, "div[data-component-type='s-search-result']")

    for product in products:

        # desc = product.find_element(By.CSS_SELECTOR, "h2.a-spacing-none.a-color-base.a-text-normal").text
        # img = product.find_element(By.CSS_SELECTOR, "img.s-image").get_attribute("src")
        # offer_price = product.find_element(By.CSS_SELECTOR, "span.a-price-whole").text
        try:
            rating = product.find_element(By.CSS_SELECTOR, "span.a-size-small.a-color-base").text
        except:
            rating = None
        print(rating)

        
```
